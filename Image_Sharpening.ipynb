```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Upload gambar ke folder workspace, lalu masukkan nama file di sini.
filename = input("Enter the name/path of the image file: ").strip()
print("Selected file:", filename)

# Read image
img = cv2.imread(filename)

if img is None:
    raise FileNotFoundError(
        f"Image not found: {filename}. "
        "Make sure the image is uploaded to the workspace folder."
    )

# Convert BGR -> RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Sharpening
kernel_sharpening = np.array([
    [-1, -1, -1],
    [-1,  9, -1],
    [-1, -1, -1]
])

sharpened = cv2.filter2D(img_rgb, -1, kernel_sharpening)

# Brightness
brightness_value = int(
    input("Brightness, example 50 for Bright or -50 for Dark: ")
)

brightened = cv2.convertScaleAbs(
    img_rgb,
    alpha=1,
    beta=brightness_value
)

# Output
plt.figure(figsize=(15, 8))

plt.subplot(1, 3, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(sharpened)
plt.title("Sharpened Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(brightened)
plt.title("Brightened Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
