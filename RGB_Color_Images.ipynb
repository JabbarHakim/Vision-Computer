```python
import cv2

filename = "/istockphoto-2201437069-612x612.jpg"

img = cv2.imread(filename)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
R, G, B = cv2.split(img)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

plt.subplot(1,4,1)
plt.imshow(img)
plt.title("Original")
plt.axis('off')

plt.subplot(1,4,2)
plt.imshow(R, cmap='gray')
plt.title("Red")
plt.axis('off')

plt.subplot(1,4,3)
plt.imshow(G, cmap='gray')
plt.title("Green")
plt.axis('off')

plt.subplot(1,4,4)
plt.imshow(B, cmap='gray')
plt.title("Blue")
plt.axis('off')

plt.show()
```
