```python
import cv2
import matplotlib.pyplot as plt

# Upload gambar ke folder workspace, lalu masukkan nama file di sini.
filename = input("Enter the name/path of the image file: ").strip()
print("Selected file:", filename)

# Read image
img = cv2.imread(filename)

if img is None:
    raise FileNotFoundError(
        f"Image not found: {filename}. "
        "Make sure the image is uploaded to the workspace folder."
    )

# Convert BGR -> RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Filtering
# Mean Filter (Average Blur)
mean_filtered = cv2.blur(img_rgb, (5, 5))

# Median Filter
median_filtered = cv2.medianBlur(img_rgb, 5)

# Gaussian Filter
gaussian_filtered = cv2.GaussianBlur(img_rgb, (5, 5), 0)

# Output
plt.figure(figsize=(15, 8))

plt.subplot(2, 2, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(mean_filtered)
plt.title("Mean Filtered Image")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(median_filtered)
plt.title("Median Filtered Image")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(gaussian_filtered)
plt.title("Gaussian Filtered Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
