```python
import cv2

filename = "/content/istockphoto-2201437069-612x612.jpg"

img = cv2.imread(filename)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
R, G, B = cv2.split(img)
```

```python
r_ = R.astype(float) / 255.0
g_ = G.astype(float) / 255.0
b_ = B.astype(float) / 255.0
```

```python
import numpy as np

K = 1 - np.max([r_,g_,b_], axis = 0)
C = (1 - r_ - K) / (1 - K + 1e-8)
M = (1 - g_ - K) / (1 - K + 1e-8)
Y = (1 - b_ - K) / (1 - K + 1e-8)
```

```python
C = (C * 255).astype(np.uint8)
M = (M * 255).astype(np.uint8)
Y = (Y * 255).astype(np.uint8)
K = (K * 255).astype(np.uint8)
```

```python
from matplotlib import cm
import matplotlib.pyplot as plt

plt.subplot(2,4,5)
plt.imshow(C, cmap="gray")
plt.title("Cyan (C)")
plt.axis("off")

plt.subplot(2,4,6)
plt.imshow(M, cmap="gray")
plt.title("Magenta (M)")
plt.axis("off")

plt.subplot(2,4,7)
plt.imshow(Y, cmap="gray")
plt.title("Yellow (Y)")
plt.axis("off")

plt.subplot(2,4,8)
plt.imshow(K, cmap="gray")
plt.title("Black (K)")
plt.axis("off")

plt.tight_layout()
plt.show()
```
